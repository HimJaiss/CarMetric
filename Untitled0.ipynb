```python
import pandas as pd
```

```python
df = pd.read_csv('Car Dataset Processed.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
df['insurance_validity'].value_counts()
```

```python
df['insurance_validity'] = df['insurance_validity'].map({'First Owner':1 , 'Second Owner':2 , 'Third Owner':3 , 'Fifth Owner':5})
df['ownsership'].value_counts()
```

```python
df['fuel_type'].value_counts()
```

```python
df = df.dropna(subset=['registration_year'])
df['registration_year'] = df['registration_year'].fillna('0')
df['registration_year'] = df['registration_year'].astype(str)
df['registration_year'] = df['registration_year'].str.split('-').str[1]
df['registration_year'] = pd.to_numeric(df['registration_year'], errors='coerce')
df = df.dropna(subset=['registration_year'])
df['registration_year'] = df['registration_year'].astype(int)
df['registration_year'] = df['registration_year'] + 2000
df['registration_year'].value_counts()
```

```python
X = df[['insurance_validity','fuel_type','kms_driven','ownsership','transmission']]
y = df['price(in lakhs)']
```

```python
X.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
```

```python
models = {'LinReg': LinearRegression(),
          'KNN-3': KNeighborsRegressor(n_neighbors=3),
          'SVM-Poly': SVR(kernel='poly')}

for name,model in models.items():
  model.fit(X, y.ravel() if len(y.shape) > 1 else y)
  sc=model.score(X,y)
  print(name,sc)
```

```python
final_model = models['KNN-3']
```

```python
d1 = {'Compregensive': 0,'Third Party insurance' : 1,'Zero Dep': 2, 'Not Availabe': 3, 'Third Party': 1}
d2 = {'Petrol': 0, 'Diesel': 1, 'CNG': 2}
d3 = {'Mannual': 0, 'Automatic': 1}
d4 = {'First Owner': 1 , 'Second Owner': 2 , 'Third Owner': 3 , 'Fifth Owner': 5, 'Fourth Owner': 4}
```

```python
insurance_validity = input('Insurance Validity:')
fuel_type = input('Fuel Type:')
kms_driven = input('Kms Driven:')
ownsership = input('Ownsership:')
transmission = input('Transmission type:')

insurance_validity = d1[insurance_validity]
fuel_type = d2[fuel_type]
transmission = d3[transmission]
ownsership = d4[ownsership]

kms_driven = int(kms_driven)

test = [[insurance_validity, fuel_type, kms_driven, ownsership, transmission]]

yp = int(final_model.predict(test)[0])
print('Predicted Price of the car is {} Rs.'.format(yp))
```
